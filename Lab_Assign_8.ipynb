```python
# Reinforcement Learning - Assignment 1
# Calculate State Value and Action Value for each state without Bellman Equation

# --- Environment Setup ---
# States are represented as (row, col) coordinates
# Grid: 2 rows x 3 cols
# S0: (0,0), S1: (0,1), G: (0,2)
# S3: (1,0), S4: (1,1), X: (1,2)

GOAL_STATE = (0, 2)
OBSTACLE_STATE = (1, 2)
DISCOUNT_FACTOR = 0.9

# Policy (Pi): [S0: Right, S1: Down, S3: Right, S4: Up]
policy = {
    (0, 0): 'Right', # S0
    (0, 1): 'Down',  # S1
    (1, 0): 'Right', # S3
    (1, 1): 'Up'     # S4
}

def step(state, action):
    """
    Given a state and an action, returns the next state, reward, and whether the episode is done.
    Transitions are deterministic.
    """
    row, col = state

    # Determine tentative next position
    if action == 'Up':
        row -= 1
    elif action == 'Down':
        row += 1
    elif action == 'Left':
        col -= 1
    elif action == 'Right':
        col += 1

    # Check boundaries (Grid is 2x3)
    if row < 0 or row >= 2 or col < 0 or col >= 3:
        # Action takes agent outside the grid, stay in same state
        next_state = state
    else:
        next_state = (row, col)

    # Determine reward and termination
    if next_state == GOAL_STATE:
        return next_state, 10, True
    elif next_state == OBSTACLE_STATE:
        return next_state, -5, True
    else:
        return next_state, -1, False

def calculate_discounted_return(start_state, start_action=None, max_steps=1000):
    """
    Calculates the return by physically following the path and accumulating discounted rewards.
    We use max_steps to prevent infinite loops (like S1 -> S4 -> S1).
    """
    current_state = start_state
    total_return = 0.0
    discount = 1.0

    for step_idx in range(max_steps):
        # Determine which action to take
        if step_idx == 0 and start_action is not None:
            action = start_action
        else:
            action = policy.get(current_state)
            if action is None:
                break # Should not happen unless policy is undefined for a state

        # Take the step
        next_state, reward, done = step(current_state, action)

        # Accumulate discounted reward
        total_return += discount * reward

        # Update discount factor for next step
        discount *= DISCOUNT_FACTOR

        if done:
            break

        current_state = next_state

    return total_return

# --- Value Functions ---

def state_value_function(state):
    """Calculates V(s) by following the policy."""
    return calculate_discounted_return(state)

def action_value_function(state, action):
    """Calculates Q(s, a) by taking action 'a' first, then following the policy."""
    return calculate_discounted_return(state, start_action=action)


# --- Execution and Output ---
if __name__ == "__main__":
    s0 = (0, 0)

    # Calculate and display State Value for S0
    v_s0 = state_value_function(s0)
    print(f"State Value for state S0: {v_s0:.4f}")

    # Calculate and display Action Value for S0 and action 'Up'
    q_s0_up = action_value_function(s0, 'Up')
    print(f"Action Value for state S0, action 'Up': {q_s0_up:.4f}")
```
