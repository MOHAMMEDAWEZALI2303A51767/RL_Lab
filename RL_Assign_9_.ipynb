```python

# Assignment 1: Bellman Expectation Equations
# Stochastic Policy and Deterministic Environment

gamma = 0.9

# Solve simultaneous equations algebraically
# V(S0) = -1 + 0.81 * V(S1)
# V(S1) = 89.9 + 0.09 * V(S0)

V_S0 = 71.819 / (1 - 0.81 * 0.09)
V_S1 = 89.9 + 0.09 * V_S0

# Calculate action values
Q_S0_Down = -1 + gamma * V_S1
Q_S0_Right = -1

Q_S1_Right = 100
Q_S1_Up = -1 + gamma * V_S0

# Display results
print("Assignment 1 Results")
print("--------------------")
print("V(S0) =", round(V_S0, 2))
print("V(S1) =", round(V_S1, 2))

print("Q(S0, Down) =", round(Q_S0_Down, 2))
print("Q(S0, Right) =", round(Q_S0_Right, 2))
print("Q(S1, Right) =", round(Q_S1_Right, 2))
print("Q(S1, Up) =", round(Q_S1_Up, 2))

# Verification
print("\nVerification")
print("V(S0) =", round(
    0.9 * Q_S0_Down + 0.1 * Q_S0_Right, 2))
print("V(S1) =", round(
    0.9 * Q_S1_Right + 0.1 * Q_S1_Up, 2))
```

```python

# Assignment 2: Stochastic Environment
# Deterministic Policy and Iterative Policy Evaluation

gamma = 0.9
theta = 0.000001

# Initialize state values
V_S0 = 0
V_S1 = 0
iterations = 0

# Iterative Policy Evaluation
while True:
    new_S0 = 0.1 * (-1) + 0.9 * (-1 + gamma * V_S1)

    new_S1 = 0.9 * (100) + 0.1 * (-1 + gamma * V_S0)

    delta = max(abs(new_S0 - V_S0),
                abs(new_S1 - V_S1))

    V_S0 = new_S0
    V_S1 = new_S1
    iterations += 1

    if delta < theta:
        break

# Calculate action values
Q_S0_Down = 0.1 * (-1) + 0.9 * (-1 + gamma * V_S1)
Q_S0_Right = -1

Q_S1_Right = 0.9 * (100) + 0.1 * (-1 + gamma * V_S0)
Q_S1_Up = -1 + gamma * V_S0

# Display results
print("Assignment 2 Results")
print("--------------------")
print("Number of iterations =", iterations)

print("V(S0) =", round(V_S0, 2))
print("V(S1) =", round(V_S1, 2))

print("Q(S0, Down) =", round(Q_S0_Down, 2))
print("Q(S0, Right) =", round(Q_S0_Right, 2))
print("Q(S1, Right) =", round(Q_S1_Right, 2))
print("Q(S1, Up) =", round(Q_S1_Up, 2))

# Verification
print("\nPolicy Verification")
print("V(S0) = Q(S0, Down):",
      round(V_S0, 2) == round(Q_S0_Down, 2))

print("V(S1) = Q(S1, Right):",
      round(V_S1, 2) == round(Q_S1_Right, 2))
```
