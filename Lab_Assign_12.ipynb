```python
import numpy as np

def value_iteration(gamma=0.9, theta=1e-6):
    # States: S0, S1, G (Goal is terminal with V(G) = 0)
    states = ['S0', 'S1', 'G']
    actions = ['Up', 'Down', 'Left', 'Right']

    # Initialize state values
    V = {'S0': 0.0, 'S1': 0.0, 'G': 0.0}

    # Transition and Reward Dynamics for Deterministic Grid World
    # Assuming standard 2x2 grid layout:
    # S0 -> Down leads to S1 (reward -1)
    # S1 -> Right leads to G (reward +100)
    # Hitting walls or invalid directions results in staying in the same state with reward -1
    def get_transition_and_reward(state, action):
        if state == 'G':
            return 'G', 0.0

        if state == 'S0':
            if action == 'Down':
                return 'S1', -1.0
            else: # Up, Left, Right result in hitting wall or staying in S0
                return 'S0', -1.0

        elif state == 'S1':
            if action == 'Right':
                return 'G', 100.0
            elif action == 'Up':
                return 'S0', -1.0
            else: # Down, Left
                return 'S1', -1.0

        return state, -1.0

    iteration = 0
    iteration_history = []

    print("--- Starting Value Iteration ---\n")

    # Value Iteration Algorithm Loop
    while True:
        iteration_history.append({'S0': V['S0'], 'S1': V['S1']})
        delta = 0
        V_new = V.copy()

        for s in ['S0', 'S1']: # Non-terminal states
            max_q = float('-inf')
            for a in actions:
                next_state, reward = get_transition_and_reward(s, a)
                # Q(s, a) = R + gamma * V(s')
                q_val = reward + gamma * V[next_state]
                if q_val > max_q:
                    max_q = q_val

            V_new[s] = max_q
            delta = max(delta, abs(V_new[s] - V[s]))

        V = V_new
        iteration += 1

        if delta < theta:
            iteration_history.append({'S0': V['S0'], 'S1': V['S1']})
            break

    # 1. Display Iteration-wise State-Value Table
    print(f"Total Iterations to Converge: {iteration}\n")
    print("Iteration-wise State-Value Table:")
    print(f"{'Iteration':<12} | {'V(S0)':<10} | {'V(S1)':<10}")
    print("-" * 38)
    for idx, vals in enumerate(iteration_history):
        print(f"{idx:<12} | {vals['S0']:<10.4f} | {vals['S1']:<10.4f}")
    print("\n" + "="*50 + "\n")

    # 2. Final Action-Value Table Q*(s, a) and Optimal Policy Calculation
    print("Final Action-Value Table Q*(s, a):")
    print(f"{'State':<8} | {'Action':<8} | {'Next State':<10} | {'Q(s, a)':<10}")
    print("-" * 42)

    optimal_policy = {}

    for s in ['S0', 'S1']:
        best_action = None
        max_q = float('-inf')
        for a in actions:
            next_state, reward = get_transition_and_reward(s, a)
            q_val = reward + gamma * V[next_state]
            print(f"{s:<8} | {a:<8} | {next_state:<10} | {q_val:<10.2f}")

            if q_val > max_q:
                max_q = q_val
                best_action = a
        optimal_policy[s] = best_action
        print("-" * 42)

    # 3. Display Final Optimal State Values
    print("\nFinal Optimal State Values V*(s):")
    for s in states:
        print(f"V*({s}) = {V[s]:.2f}")

    # 4. Display Optimal Policy
    print("\nOptimal Policy:")
    for s in ['S0', 'S1']:
        print(f"Optimal action for {s}: {optimal_policy[s]}")

if __name__ == "__main__":
    value_iteration()
```
